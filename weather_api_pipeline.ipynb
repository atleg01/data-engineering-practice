```python
import requests
import pandas as pd
url = "https://api.open-meteo.com/v1/forecast?latitude=-25.75&longitude=28.23&daily=temperature_2m_max,temperature_2m_min,precipitation_sum&timezone=Africa/Johannesburg"
```

```python
response = requests.get(url, headers={"Accept": "application/json"})

data = response.json()
print(data)
```

```python
df = pd.DataFrame(data['daily'])
print(df.head())
print(df.info())
print(df.isnull().sum())
```

```python
payload = {
    "name": "Atlegang",
    "course": "Data Engineering",
    "progress": "API module complete"
}

response = requests.post("https://httpbin.org/post", json=payload)
result = response.json()
print(result['json'])
```
